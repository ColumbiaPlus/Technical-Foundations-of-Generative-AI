**Input size of a text neural network**
* In a basic neural network, words are entered one by one as input to the network
* Each input is the size of the vocabulary of the problem being studied
* If the text consists of 1000 sentences, each exactly 10 words in length, and if the vocabulary is 1000 words, then the size of the input to the neural network is:
     1000 (sentences) X 10 (words) X 1000 (length of each vector)
     = 10,000,000 (10 million)
* If the vocabulary is 1 million words (the approximate vocabulary of the English language), then the input size is:
     1000 (sentences) X 10 (words) X 1000000 (length of each vector)
     = 10,000,000,000 (10 billion)
* This is a lot! A bigger input layer results in a much larger network and a large network needs far greater processing power
* If we modify the network to introduce one sentence at a time, with a million vocabulary the size would be:
     1000 (sentences) * 1000000 (length of each vector)
     
     = 1,000,000,000 (1 billion)
* Which is still a huge network!


**Embedding**
* Embedding is used to decrease the size of the input while still retaining additiona linformation
* The idea is to convert each word into a vector and convert the entire input into a matrix of order

   sentences X words X embedding_dimension

* For example, if we have 1000 sentences, each exactly 10 words in length, and use an embedding vector of size 20, then the input dimension is:

     1000 X 10 X 20

* And the size of the input is 200,000 (compare this with the un-embedded example)

* **The size of the input is independent of the size of the vocabulary!**
* **And, words that are closer in meaning may have vectors that are closer together (assuming you have enough data)**

**Example**

* Each sentence is a vector containing the integer value of the word in the vocabulary

** Embedding
* The code below creats an embedding vectors for the same sentences
* Each word has an embedding vector of length 32 associated with it
* Each sentence has 10 words, so we have 10 * 32 data requirement for each sentence
* There are 1,000 sentences, so our total data requirement is 1000 * 10 * 32 or 320,000
* **Note that this is independent of the length of tha vocabulary**
* Whether the vocabulary is 1,000 or 1,000,000, the data requirements for the embedding stay the same
* The size of the input layer of the neural network is the size of the embedding vector (32 in our example)

```python
import tensorflow as tf
import numpy as np

k = 1000 #Number of sentences
m = 10 #Number of words
v = 100000 #Number of words
e = 32 #Length of embedding vector

#Input data
# Create the input data (1000 sentences of 10 words each)
#v is the vocabulary size. Each word is a number between 0 and v-1
#k is the number of sentences
#m is the number of words in each sentence
input_data = np.random.randint(v, size=(k, m))


#Create the embedding vectors

model = tf.keras.Sequential()
model.add(tf.keras.layers.Embedding(v, e, input_length=m))

#Create output_array that contains the embedding
model.compile('rmsprop', 'mse')
output_array = model.predict(input_data)
print(output_array.shape)
print("Size of the output: ",output_array.size)
```
